# PDF Based Chatbot

```python
from langchain_community.document_loaders import PyPDFLoader

loader = PyPDFLoader("Resume -Mitesh Upadhyay.pdf")
```

```python
documents = loader.load()
context = "\n\n".join(doc.page_content for doc in documents)
```

```python
from langchain_core.prompts import PromptTemplate

prompt = PromptTemplate(
    input_variables=["context", "question"],
    template=""" 
    You are an advance resume evaluation assitant. Analyze the provided resume.
    score it out of 100 based on the following criteria: Clarity, relevance, format, comprehensiveness.

    Your response should be in the following format:
    1. **Score**: Provide the score out of 100.
    2. **Strengths**: List atleast three strengths found in the resume.
    3. **Weaknesses**: List atleast three weaknesses or areas of improvement found in the resume.
    4.**Skills mentioned**: Identify and list the skills that are explicitly mentioned in the resume.
    5. **Recommended Skills**: Suggest additional skills that could enhance the resume based on the job market trends and industry requirements.
    6. **Next Career Path**: Provide a brief recommendation for the next career path or role that aligns with the individual's experience and skills.


    Please ensure your analysis is clear and concise, providing actionable feedback for the individual to improve their resume.
    """
)
```

```python
question = "Please evaluate this resume"

formatted_prompt = prompt.format(context=context, question=question)

response = llm.invoke(formatted_prompt)

print(response.content)
```
